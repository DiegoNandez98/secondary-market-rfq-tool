# Predictive Flow Integration — TBD

The current RFQ tool prioritises requests once they arrive.

The retail-flow model adds a complementary layer by estimating the probability that an existing client position generates a SELL within the next 24 hours.

A natural next step would be to connect both components.

For SELL RFQs, the latest position-level `sell_probability` could be matched to the incoming RFQ using client and product information and used as an additional prioritisation signal.

The same probabilities could also be used before RFQs arrive to estimate expected secondary-market SELL flow given current market conditions.

Possible extensions:

- add `sell_probability` to the RFQ prioritisation score
- aggregate position-level probabilities into expected sell flow
- retrain the model regularly on new (real) transaction data
- build a separate BUY-flow model to estimate both sides of expected secondary-market activity
- test whether predicted flow improves RFQ prioritisation on historical or live data

This integration is left as future work.
