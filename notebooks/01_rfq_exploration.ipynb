We want to check now: how old is our current price with respect to the current market? 
We look at:
1. how old is our price
2. how did the underlying move recently
3. how did IV change recently

```python

def add_staleness_risk(df):

    # work on a copy of the input DataFrame
    df = df.copy()

    # assume 15min old price is already too old, i.e. max score
    df ["age_score"] = (df["time_since_last_price_update_min"] / 15).clip(0,1)

    # assume that if underlying moves 2% in the last 5 min, has maximum score
    df["recent_move_score"] = (df["underlying_move_5m_pct"].abs()/2).clip(0,1)

    # assume a change of 3 vol points has maximum score
    df ["recent_iv_score"] = (df["iv_change_5m"].abs()/3).clip(0,1)

    # combine underlying and iv movements
    # weights are assigned heuristically, with real desk data you calibrate them empirically
    df["recent_market_score"] = (0.7*df["recent_move_score"] + 0.3 * df["recent_iv_score"])

    # final score
    df["staleness_risk"] = (df["age_score"]* (0.3 + 0.7 * df["recent_market_score"]))

    return df
```

The initial weights are heuristic because we have no historical desk data. 
In production we can define an observable target, for example:
- repricing error for staleness
- cost of delay for RFQ priority 

and calibrate the weights using historical observations, and use a model such as logistic or linear regression and out-of-sample validation.

```python
df = add_staleness_risk(df)

df [["underlying", "time_since_last_price_update_min", 
     "underlying_move_5m_pct","iv_change_5m", 
     "staleness_risk"]].sort_values("staleness_risk", ascending = 0)
```

Next step: build overall priority score

```python
df.columns
```

```python


# assume 2M or more has maximum score
df["size_score"] = (df["notional"]/2_000_000).clip(0,1)

# assume barrier distance 30% has risk score 0
df["barrier_risk_score"] = (1 - df["distance_to_barrier_pct"]/30).clip(0,1).fillna(0)

# Assume 5% daily move or more = maximum market move score
# here in reality it should depend on the typical daily volatility
df["market_move_score"] = ( df["abs_underlying_move_pct"] / 5).clip(0, 1)

# Assume 8+ vol points change = maximum IV score
df["iv_score"] = (df["abs_iv_change"] / 8).clip(0, 1)

```

```python
df["priority_score"] = (
    0.35 * df["barrier_risk_score"]
    + 0.20 * df["market_move_score"]
    + 0.15 * df["iv_score"]
    + 0.15 * df["staleness_risk"]
    + 0.15 * df["size_score"]
)

df[[
    "client",
    "product_type",
    "underlying",
    "direction",
    "notional",
    "priority_score"
]].sort_values(
    "priority_score",
    ascending=False
)
```
